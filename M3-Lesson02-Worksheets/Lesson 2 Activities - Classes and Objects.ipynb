## Activity 4
Create a method for your `Noun` class. The method should perform a specific activity (or series of activities) on objects based on the class. 

Although this can be a simple `display`, you are encouraged to consider more complicated activities in the method.

```python
# your code here

class Laptop:
    def __init__(self, brand, model, screen_size, operating_system):
        self.brand = brand
        self.model = model
        self.screen_size = screen_size
        self.operating_system = operating_system

    def start(self):
        print(f"{self.brand} {self.model} is starting.")
        print(f"Loading {self.operating_system}...")
        print("The laptop is ready to use.")


laptop1 = Laptop("Dell", "XPS 13", 13.4, "Windows")
laptop2 = Laptop("Apple", "MacBook Air", 13.6, "macOS")

laptop1.start()
print()
laptop2.start()
```

## Activity 5
Create a separate class that is logically related to your original `Noun` class, based on the example that a `Person` can have one or more `Addresses`.
	
Create at least one `Noun` object that includes a list of objects from the associated class. Display the `Noun` with the related objects.

```python
# your code here
class Accessory:
    def __init__(self, name, accessory_type):
        self.name = name
        self.accessory_type = accessory_type

    def display(self):
        return f"{self.name} ({self.accessory_type})"


class Laptop:
    def __init__(
        self,
        brand,
        model,
        screen_size,
        operating_system,
        accessories=None
    ):
        self.brand = brand
        self.model = model
        self.screen_size = screen_size
        self.operating_system = operating_system
        self.accessories = accessories if accessories is not None else []

    def display(self):
        print(f"Brand: {self.brand}")
        print(f"Model: {self.model}")
        print(f"Screen size: {self.screen_size} inches")
        print(f"Operating system: {self.operating_system}")
        print("Accessories:")

        for accessory in self.accessories:
            print(f"- {accessory.display()}")


wireless_mouse = Accessory("Wireless Mouse", "Input Device")
laptop_stand = Accessory("Adjustable Laptop Stand", "Stand")
usb_hub = Accessory("USB-C Hub", "Adapter")

laptop = Laptop(
    "Dell",
    "XPS 13",
    13.4,
    "Windows",
    [wireless_mouse, laptop_stand, usb_hub]
)

laptop.display()
```
