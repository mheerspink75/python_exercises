## Practice 4
Create a `display_info` method for the `Animal` class. 

Create three objects from the `Animal` class and display each object using the `display_info` method.

```python
# your code here
class Animal:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Species: {self.species}")
        print(f"Age: {self.age}")
        print()


# Create three Animal objects
dog = Animal("Buddy", "Dog", 3)
cat = Animal("Whiskers", "Cat", 5)
bird = Animal("Tweety", "Bird", 1)

# Display information for each animal
dog.display_info()
cat.display_info()
bird.display_info()
```

```python
# your code here
class Animal:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Species: {self.species}")
        print(f"Age: {self.age}")

    def make_sound(self):
        print(f"{self.name} the {self.species} makes a sound!")


# Create an object
dog = Animal("Buddy", "Dog", 3)

# Execute the method
dog.make_sound()
```
