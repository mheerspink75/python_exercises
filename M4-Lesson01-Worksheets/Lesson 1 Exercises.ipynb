## Activity 1
Write a program to display the following: 

* Current date and time 
* Current year 
* Current month
* Week number of the year 
* Weekday of the week 
* Day of the year 
* Day of the month 
* Day of the week 

```python
# your code here
import datetime

current_datetime = datetime.datetime.now()

# Get the current date and time
print("Current date and time:", current_datetime)

# Get the current year
print("Current year:", current_datetime.year)

# Get the current month
print("Current month:", current_datetime.month)

# Get the current week number of the year
print("Week number of the year:", current_datetime.isocalendar().week)

# Get the current weekday of the week
print("Weekday of the week:", current_datetime.strftime("%A"))

# Get the current day of the year
print("Day of the year:", current_datetime.timetuple().tm_yday)

# Get the current day of the month
print("Day of the month:", current_datetime.day)

# Get the current day of the week
print("Day of the week:", current_datetime.strftime("%A"))
```

## Activity 2 

Write a function that prompts the user for a year and checks if it is a leap year. The function should return `True` if the input is a leap year and `False` otherwise. 

Make sure to test your code with various inputs. 

```python
# your code here
def check_leap_year():
    year = int(input("Enter a year: "))
    return year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)

print("Is it a leap year?", check_leap_year())
```

## Activity 3 

Write a function that prompts the user for a date (or uses the current date if the user does not input a date) and subtracts five days from that date. The function should return the new date.

Display the result to the user. 

```python
# your code here
from datetime import datetime, timedelta

def subtract_five_days():
    date_input = input("Enter a date in YYYY-MM-DD format (or press Enter for today): ").strip()

    if date_input == "":
        selected_date = datetime.now().date()
    else:
        selected_date = datetime.strptime(date_input, "%Y-%m-%d").date()

    new_date = selected_date - timedelta(days=5)
    return new_date

result = subtract_five_days()
print("Five days earlier:", result)
```

## Activity 4 

Write a program that converts a Unix timestamp string to a human-readable date. 

```python
# your code here 
# Timestamp - 1788039735
import datetime

def convert_unix_timestamp():
    timestamp = int(input("Enter a Unix timestamp: "))
    return datetime.datetime.fromtimestamp(timestamp).strftime("%Y-%m-%d %H:%M:%S")

result = convert_unix_timestamp()
print("Human-readable date:", result)
```

## Activity 5 

Write a program that prints the dates for yesterday, today, and tomorrow. 

```python
# your code here
from datetime import date, timedelta

today = date.today()
yesterday = today - timedelta(days=1)
tomorrow = today + timedelta(days=1)

print("Yesterday:", yesterday)
print("Today:", today)
print("Tomorrow:", tomorrow)
```

## Activity 6 

Write a program that prints the date for the next five days, starting from today. 

```python
# your code here
from datetime import date, timedelta

today = date.today()
for i in range(1, 6):
    next_day = today + timedelta(days=i)
    print("Day", i, ":", next_day)
```

## Activity 7 

Write a program that adds five seconds to the current time and displays the result. 

```python
# your code here
# Write a program that adds five seconds to the current time and displays the result. 
from datetime import datetime, timedelta

current_time = datetime.now()
new_time = current_time + timedelta(seconds=5)
print("Current time:", current_time.strftime("%Y-%m-%d %H:%M:%S"))
print("Time after adding 5 seconds:", new_time.strftime("%Y-%m-%d %H:%M:%S"))
```

## Activity 8 

Write a program that prompts the user for two different dates, computes the number of days between those dates, and displays the results. 

The order in which the user enters the dates should not affect the results. 

```python
# your code here
from datetime import datetime

date_format = "%Y-%m-%d"
date1 = input("Enter the first date (YYYY-MM-DD): ")
date2 = input("Enter the second date (YYYY-MM-DD): ")

d1 = datetime.strptime(date1, date_format)
d2 = datetime.strptime(date2, date_format)

delta = abs(d2 - d1)
print("Number of days between the dates:", delta.days)
```
