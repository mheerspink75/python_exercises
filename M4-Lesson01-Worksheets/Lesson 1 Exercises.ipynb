## Activity 1
Write a program to display the following: 

* Current date and time 
* Current year 
* Current month
* Week number of the year 
* Weekday of the week 
* Day of the year 
* Day of the month 
* Day of the week 

```python
# your code here
import datetime

current_datetime = datetime.datetime.now()

# Get the current date and time
print("Current date and time:", current_datetime)

# Get the current year
print("Current year:", current_datetime.year)

# Get the current month
print("Current month:", current_datetime.month)

# Get the current week number of the year
print("Week number of the year:", current_datetime.isocalendar().week)

# Get the current weekday of the week
print("Weekday of the week:", current_datetime.strftime("%A"))

# Get the current day of the year
print("Day of the year:", current_datetime.timetuple().tm_yday)

# Get the current day of the month
print("Day of the month:", current_datetime.day)

# Get the current day of the week
print("Day of the week:", current_datetime.strftime("%A"))
```

## Activity 2 

Write a function that prompts the user for a year and checks if it is a leap year. The function should return `True` if the input is a leap year and `False` otherwise. 

Make sure to test your code with various inputs. 

```python
# your code here
def check_leap_year():
    year = int(input("Enter a year: "))

    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False


print("Is it a leap year?", check_leap_year())
```
