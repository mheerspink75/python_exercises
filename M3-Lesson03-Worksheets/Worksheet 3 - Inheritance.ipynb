```python
class Person:
    first_name = ""
    last_name = ""

    def __init__(self, fname, lname):
        self.first_name = fname
        self.last_name = lname

    def display(self):
        print("Person First Name: " + self.first_name)
        print("Person Last Name: " + self.last_name)


class Employee(Person):
    employee_id = None

    def __init__(self, emp_id, fname, lname):
        super().__init__(fname, lname)
        self.employee_id = emp_id

    def display(self):
        print("Employee ID: " + self.employee_id)
        print("Employee First Name: " + self.first_name)
        print("Employee Last Name: " + self.last_name)


class Director(Employee):
    director_level = None
    team = None

    def __init__(self, emp_id, fname, lname, d_level):
        Employee.__init__(self, emp_id, fname, lname)
        self.director_level = d_level
        self.team = list()

    def display(self):
        print("Director Information:")
        print("Employee ID: " + self.employee_id)
        print("First Name: " + self.first_name)
        print("Last Name: " + self.last_name)
        print("Director Level: " + self.director_level)
        print("Team Members:")

        for employee in self.team:
            employee.display()
            print()

    def add_employee(self, emp):
        self.team.append(emp)


class HourlyEmployee(Employee):
    hourly_rate = None

    def __init__(self, emp_id, fname, lname, hourly_rate=25.00):
        super().__init__(emp_id, fname, lname)
        self.hourly_rate = hourly_rate

    def display(self):
        super().display()
        print(f"Hourly Rate: ${self.hourly_rate:.2f}")


d1 = Director("E24523525", "Haythem", "Balti", "D-LEVEL-1")

e1 = Employee("E4746456456", "Mark", "Smith")
d1.add_employee(e1)

e2 = HourlyEmployee("E47464578978", "Mary", "Lang")
d1.add_employee(e2)

d1.display()
```

Implement the classes below as follows:
* Implement the `__init__` and `display` methods of the `Asset` class.
* Next, leverage the `__init__` and `display` methods from the `Asset` class to create the following:
    * Implement the `__init__` and `display` methods of the `Stock` class. 
    * Implement the `__init__` and `display` methods of the `Bond` class. 
    * Implement the `__init__` and `display` methods of the `MutualFund` class. 
* Create two objects from each of the `Stock`, `Bond`, and `MutualFund` classes. Use the `display` method to display the contents of each object.  

```python
class Asset:
    symbol = None
    price = None
    description = None

    def __init__(self, symbol, price, description):
        self.symbol = symbol
        self.price = price
        self.description = description

    def display(self):
        print(f"Symbol: {self.symbol}")
        print(f"Price: ${self.price:.2f}")
        print(f"Description: {self.description}")


class Stock(Asset):
    company_name = None

    def __init__(self, symbol, price, description, company_name):
        super().__init__(symbol, price, description)
        self.company_name = company_name

    def display(self):
        super().display()
        print(f"Company Name: {self.company_name}")


class Bond(Asset):
    interest_rate = None

    def __init__(self, symbol, price, description, interest_rate):
        super().__init__(symbol, price, description)
        self.interest_rate = interest_rate

    def display(self):
        super().display()
        print(f"Interest Rate: {self.interest_rate}%")


class MutualFund(Asset):
    fund_manager = None

    def __init__(self, symbol, price, description, fund_manager):
        super().__init__(symbol, price, description)
        self.fund_manager = fund_manager

    def display(self):
        super().display()
        print(f"Fund Manager: {self.fund_manager}")


stock1 = Stock("AAPL", 225.50, "Technology company stock", "Apple Inc.")
stock2 = Stock("MSFT", 510.25, "Software company stock", "Microsoft Corporation")

bond1 = Bond("US10Y", 98.75, "Ten-year government bond", 4.25)
bond2 = Bond("CORP01", 101.20, "Corporate bond", 5.10)

fund1 = MutualFund("VFIAX", 485.30, "Large-cap index fund", "Jane Williams")
fund2 = MutualFund("FXAIX", 212.80, "S&P 500 index fund", "Robert Davis")

stock1.display()
print()
stock2.display()
print()

bond1.display()
print()
bond2.display()
print()

fund1.display()
print()
fund2.display()
```
